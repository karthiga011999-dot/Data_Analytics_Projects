```python
#Price Analysing
print("Average Selling Price:", df["Selling Price"].mean())
print("Minimum Selling Price:", df["Selling Price"].min())
print("Maximum Selling Price:", df["Selling Price"].max())

# Average selling price by brand
brand_price = df.groupby("Brands")["Selling Price"].mean()

print(brand_price.sort_values(ascending=False).head(10))

# Top 10 brands by average selling price

brand_price.sort_values(ascending=False).head(10).plot(
    kind="pie",
    figsize=(8, 8),
    autopct="%1.1f%%"
)

plt.title("Top 10 Brands by Average Selling Price")
plt.xlabel("Brand")
plt.ylabel(" ")

plt.show()
```

```python
#Discount Analysising
print("Average Discount:", df["Discount"].mean())
print("Maximum Discount:", df["Discount"].max())

print("Average Discount Percentage:",
      df["discount percentage"].mean())

# Top 10 products with highest discount
top_discount = df.nlargest(10, "Discount")

print(top_discount[
    ["Brands", "Models", "Selling Price",
     "Original Price", "Discount"]
])
# Top 10 products by discount percentage
top_discount_percent = df.nlargest(
    10, "discount percentage"
)

top_discount_percent.plot(
    x="Models",
    y="discount percentage",
    kind="line",
    marker="o",
    figsize=(10, 5)
)

plt.title("Top 10 Products by Discount Percentage")
plt.xlabel("Mobile Model")
plt.ylabel("Discount Percentage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

2. DATA QUALITY ASSESSMENT AND PRE-PROCESSING:

```python
#Dupilicate Records
duplicates = df.duplicated().sum()
print("Duplicate Rows:", duplicates)

# Remove duplicate rows
df = df.drop_duplicates()
print("Duplicates removed successfully!")
```

```python
# Check missing values
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values[missing_values > 0])

# Display total missing cells
print("Total Missing Cells:", df.isnull().sum().sum())
```

```python
# Price Validation
negative_discount = df[df["Discount"] < 0]
print("Negative Discount Records:", len(negative_discount))

# Check selling price greater than original price
invalid_price = df[df["Selling Price"] > df["Original Price"]]
print("Invalid Price Records:", len(invalid_price))
```

```python
# Discount Accuracy
expected_discount = df["Original Price"] - df["Selling Price"]

# Check discount mismatches
discount_mismatch = df[df["Discount"] != expected_discount]

print("Discount Mismatch Records:", len(discount_mismatch))
```

3. DESCRIPTIVE PRICING ANALYTICS:

```python

#Average Selling Price
average_price = df["Selling Price"].mean()

plt.bar(["Average Selling Price"], [average_price])
plt.title("Average Selling Price of Smartphones")
plt.ylabel("Price (₹)")
plt.text(0, average_price, f"₹{average_price:,.0f}", ha="center", va="bottom")
plt.show()
```

```python
#Median Selling Price
median_price = df["Selling Price"].median()

plt.figure(figsize=(8, 5))

plt.hist(df["Selling Price"].dropna(), bins=20, edgecolor="black")

plt.axvline(
    median_price,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Median Price: ₹{median_price:,.0f}"
)

plt.title("Median Selling Price Distribution")
plt.xlabel("Selling Price (₹)")
plt.ylabel("Number of Smartphones")
plt.legend()
plt.show()
```

```python
#Original Price and Discount: 

average_original_price = df["Original Price"].mean()
average_discount = df["Discount"].mean()

labels = ["Original Price", "Average Discount"]
values = [average_original_price, average_discount]

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, values)

plt.title("Average Original Price vs Average Discount")
plt.ylabel("Amount (₹)")

plt.bar_label(bars, fmt="₹%.0f")
plt.show()
```

```python
# Mean and Median
mean_price = df["Selling Price"].mean()
median_price = df["Selling Price"].median()

print("Mean Selling Price: ₹", round(mean_price, 2))
print("Median Selling Price: ₹", round(median_price, 2))

# Bar Chart
labels = ["Mean Price", "Median Price"]
values = [mean_price, median_price]

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, values)

plt.title("Mean vs. Median Selling Price")
plt.ylabel("Selling Price (₹)")
plt.bar_label(bars, fmt="₹%.0f")
plt.show()
```

4. BRAND PORTFOLIO AND COMPETITIVE POSITIONING:

```python
# Count smartphone listings for each brand
brand_counts = df["Brands"].str.upper().value_counts()
samsung_count = brand_counts.get("SAMSUNG", 0)

print("Samsung Listings:", samsung_count)

# Create bar chart
plt.figure(figsize=(7, 5))
bars = plt.bar(["SAMSUNG"], [samsung_count], color="skyblue")

plt.title("Samsung Smartphone Listings")
plt.xlabel("Brand")
plt.ylabel("Number of Listings")
plt.bar_label(bars)

plt.show()
```

```python
# Calculate average selling price for each brand
brand_avg_price = df.groupby("Brands")["Selling Price"].mean().sort_values()

# Create horizontal bar chart
plt.figure(figsize=(10, 6))
bars = plt.barh(
    brand_avg_price.index,
    brand_avg_price.values,
    color="orange"
)

plt.title("Average Selling Price by Brand")
plt.xlabel("Average Selling Price (₹)")
plt.ylabel("Brands")

plt.bar_label(bars, fmt="₹%.0f", padding=3)
plt.tight_layout()
plt.show()
```

```python
# Compare brands by price range
brand_comparison = df.groupby("Brands")["Selling Price"].agg(
    ["min", "max", "mean"]
).round(2)

print(brand_comparison)

# Visualize minimum and maximum selling prices
brand_comparison[["min", "max"]].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Brand Comparison by Selling Price Range")
plt.xlabel("Brand")
plt.ylabel("Selling Price (₹)")
plt.legend(["Minimum Price", "Maximum Price"])
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Calculate listing count and average selling price by brand
brand_listings = df["Brands"].value_counts()
brand_avg_price = df.groupby("Brands")["Selling Price"].mean()

# Combine the results
sales_comparison = pd.DataFrame({
    "Listing Count": brand_listings,
    "Average Selling Price": brand_avg_price
}).fillna(0)

print(sales_comparison.round(2))

# Plot listing count by brand
plt.figure(figsize=(10, 5))
sales_comparison["Listing Count"].sort_values().plot(
    kind="barh", color="skyblue"
)

plt.title("Brand Listing Count Comparison")
plt.xlabel("Number of Listings")
plt.ylabel("Brand")
plt.tight_layout()
plt.show()
```

5. CUSTOMER RATING AND PRODUCT PERCEPTION:

```python
# Average Customer Rating
average_rating = df["Rating"].mean()

print("Average Customer Rating:", round(average_rating, 2), "/ 5")

# Create chart
plt.figure(figsize=(6, 4))
plt.bar(["Average Rating"], [average_rating], color="skyblue")
plt.ylim(0, 5)
plt.ylabel("Rating (out of 5)")
plt.title("Average Customer Rating")
plt.text(0, average_rating + 0.1, f"{average_rating:.2f}/5", ha="center")
plt.show()
```

```python
# Calculate average rating for brands with at least 20 rated listings
brand_rating = df.groupby("Brands")["Rating"].agg(["mean", "count"])
brand_rating = brand_rating[brand_rating["count"] >= 20]
brand_rating = brand_rating.sort_values("mean")

# Display the highest-rated brand
print(brand_rating.sort_values("mean", ascending=False))

# Create chart
plt.figure(figsize=(10, 6))
plt.barh(brand_rating.index, brand_rating["mean"], color="orange")
plt.xlabel("Average Rating")
plt.ylabel("Brand")
plt.title("Average Customer Rating by Brand")
plt.xlim(0, 5)
plt.tight_layout()
plt.show()
```

```python
# Identify pricing exceptions
negative_discount = df["Discount"] < 0
selling_above_original = df["Selling Price"] > df["Original Price"]

exception_counts = [
    negative_discount.sum(),
    selling_above_original.sum()
]

labels = [
    "Negative Discount",
    "Selling Price Above Original Price"
]

print("Negative Discount Records:", negative_discount.sum())
print("Selling Price Above Original Price:", selling_above_original.sum())

# Create chart
plt.figure(figsize=(8, 5))
bars = plt.bar(labels, exception_counts, color=["red", "steelblue"])

plt.title("Pricing Exceptions")
plt.ylabel("Number of Records")
plt.bar_label(bars)
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```
