1.Write a Python program to calculate the area of a rectangle using user input for length and width.

```python
length = float(input("Enter the length: "))
width = float(input("Enter the width: "))

area = length * width

print(f"The area of the rectangle is {area}")
```

2.Write a Python program to check whether a given number is positive, negative, or zero.

```python
num = float(input("Enter a number: "))

if num > 0:
    print(f"{num} is Positive")
elif num < 0:
    print(f"{num} is Negative")
else:
    print(f"{num} is Zero")
```

3.Write a Python program to find the largest number among three numbers entered by the user.

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
num3 = float(input("Enter the third number: "))

largest = max(num1, num2, num3)

print(f"The largest number is {largest}")
```

4.Write a Python program to count the number of vowels in a given string.

```python
text = input("Enter a string: ")

count = 0

for char in text:
    if char.lower() in "aeiou":
        count += 1

print(f"The number of vowels is {count}")
```

5.Write a Python program to generate the multiplication table of a given number (up to 10).

```python
num = int(input("Enter a number: "))

for i in range(1, 11):
    result = num * i
    print(f"{num} x {i} = {result}")
```
