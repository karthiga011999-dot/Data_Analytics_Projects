1.Write a Python program to find the factorial of a number using a loop.

```python
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num + 1):
    factorial = factorial * i

print("Factorial =", factorial)
```

2.Write a Python program to check whether a number is a palindrome.

```python
num = input("Enter a number: ")

if num == num[::-1]:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

3.Write a Python program to reverse a string entered by the user.

```python
text = input("Enter a string: ")

reverse = text[::-1]

print("Reversed string =", reverse)
```

4.Write a Python program to find the sum of digits of a number. 

```python
num = int(input("Enter a number: "))

sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit
    num = num // 10

print("Sum of digits =", sum)
```

5.Write a Python program to find the largest number in a list.

```python
numbers = [10, 25, 7, 45, 18]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest number =", largest)
```
