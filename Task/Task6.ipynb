```python
import pandas as pd

data = {
    "Name": ["A", "B", "C", "D"],
    "Marks": [80, 65, 90, 50]
}
```

```python
df = pd.DataFrame(data)

print(df)
```

```python
print(df.head(2))
```

```python
print(df.columns)
```

```python
average = df["Marks"].mean()

print("Average Marks:", average)
```

```python
import pandas as pd

data = {
    "Name": ["A", "B", "C", "D"],
    "Marks": [80, 65, 90, 50]
}

df = pd.DataFrame(data)

print("DataFrame:")
print(df)

print("\nFirst 2 rows:")
print(df.head(2))

print("\nColumn names:")
print(df.columns)

print("\nAverage Marks:")
print(df["Marks"].mean())
```
