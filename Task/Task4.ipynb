String Data Analysis - text = "Data analytics is powerful and data driven"
Count total words
Count frequency of each word
Find most repeated word

```python
text = "Data analytics is powerful and data driven"

words = text.split()

print("Total words:", len(words))
```

```python
text = "Data analytics is powerful and data driven"

words = text.lower().split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
most_repeated = max(frequency, key=frequency.get)

print("Most repeated word:", most_repeated)
```

Whole Program:

```python
text = "Data analytics is powerful and data driven"

words = text.lower().split()

print("Total words:", len(words))

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] = frequency[word] + 1
    else:
        frequency[word] = 1

print("Word frequency:", frequency)

most_repeated = max(frequency, key=frequency.get)

print("Most repeated word:", most_repeated)
```
